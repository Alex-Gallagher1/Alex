```python
#File: hw11.ipynb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
im_data = pd.read_csv("Data/astro_image (1).csv", header = None, sep = "\s+", dtype = float)
im_data_arr = im_data.to_numpy()
```

```python
#3.1 Plotting an Astronomical Image
plt.figure()

im = plt.imshow(im_data_arr, cmap = "viridis", vmin = 100, vmax = 350)
plt.colorbar(im)

plt.show()
```

```python
#3.1 Plotting an Astronomical Image
im_data1 = im_data_arr[200:400, 1000:1400]
im_data2 = im_data_arr[800:1000, 1000:1200]

fig, axs = plt.subplots(1, 2, figsize = (10,5))

im1 = axs[0].imshow(im_data1, cmap = 'viridis', vmin = 100, vmax = 350)
axs[1].imshow(im_data2, cmap = 'viridis', vmin = 100, vmax = 350)

plt.colorbar(im1)

plt.show()
```
